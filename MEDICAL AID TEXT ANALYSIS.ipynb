```python

# Make a copy of the original dataset
df_clean = df.copy()

# Remove rows where the original review is missing
df_clean = df_clean.dropna(subset=['review'])

# Fill missing cleaned reviews using the original review
df_clean['review_cleaned'] = df_clean['review_cleaned'].fillna(
    df_clean['review']
)

# Check missing values after cleaning
print("Missing values after cleaning:")
print(df_clean[['review', 'review_cleaned']].isnull().sum())

# Display the new dataset size
print("\nDataset size after cleaning:", df_clean.shape)

# Display the first five reviews
display(df_clean[['rating', 'review', 'review_cleaned']].head())
```

```python

# STEP 11: IDENTIFYING COMMON CUSTOMER CONCERNS

import pandas as pd
import re
import matplotlib.pyplot as plt
from collections import Counter

# 1. Get customer reviews
reviews = df_clean['review_cleaned'].dropna().astype(str)

# 2. Combine all reviews
all_reviews = ' '.join(reviews)

# 3. Convert text to lowercase
all_reviews = all_reviews.lower()

# 4. Remove punctuation and numbers
all_reviews = re.sub(r'[^a-z\s]', ' ', all_reviews)

# 5. Split text into individual words
words = all_reviews.split()

# 6. Remove common English words
stop_words = {
    'the', 'and', 'is', 'in', 'it', 'to', 'of', 'a',
    'for', 'this', 'that', 'was', 'with', 'on', 'my',
    'i', 'you', 'me', 'be', 'are', 'have', 'has',
    'had', 'but', 'not', 'so', 'as', 'at', 'an',
    'app', 'apps', 'very', 'just', 'can', 'use',
    'using', 'get', 'got', 'one', 'really', 'would'
}

filtered_words = [
    word for word in words
    if word not in stop_words and len(word) > 2
]

# 7. Count word frequency
word_counts = Counter(filtered_words)

# 8. Display the top 20 words
top_words = word_counts.most_common(20)

print("TOP 20 MOST FREQUENT WORDS:")

for word, count in top_words:
    print(word, ":", count)

# 9. Create a frequency chart
words_df = pd.DataFrame(
    top_words,
    columns=['Word', 'Frequency']
)

plt.figure(figsize=(12, 7))

plt.barh(
    words_df['Word'][::-1],
    words_df['Frequency'][::-1],
    color='#d63384'
)

plt.title('Top 20 Most Frequently Used Words')
plt.xlabel('Number of Occurrences')
plt.ylabel('Words')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 10. Create a word-cloud-style visualisation
# using only Matplotlib
plt.figure(figsize=(12, 7))

plt.text(
    0.5, 0.5,
    '   '.join([word for word, count in top_words]),
    fontsize=16,
    ha='center',
    va='center',
    wrap=True,
    color='#c71585'
)

plt.title('Common Words in Mental Health App Reviews')
plt.axis('off')
plt.show()
```

```python

# STEP 13: TRAINING A MACHINE LEARNING MODEL

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# 1. Prepare the dataset
model_data = df_clean[['review_cleaned', 'sentiment']].copy()

# Remove missing and empty reviews
model_data = model_data.dropna(
    subset=['review_cleaned', 'sentiment']
)

model_data['review_cleaned'] = (
    model_data['review_cleaned'].astype(str).str.strip()
)

model_data = model_data[
    model_data['review_cleaned'] != ''
]

# 2. Define input features and target variable
X = model_data['review_cleaned']
y = model_data['sentiment']

print("Total reviews used:", len(model_data))
print("\nSentiment categories:")
print(y.value_counts())

# 3. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining reviews:", len(X_train))
print("Testing reviews:", len(X_test))

# 4. Build the machine learning pipeline
model = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=20000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95,
        stop_words='english'
    )),

    ('classifier', LogisticRegression(
        max_iter=1000,
        class_weight='balanced',
        random_state=42
    ))
])

# 5. Train the model
print("\nTraining the Logistic Regression model...")

model.fit(X_train, y_train)

print("Model training completed successfully!")

# 6. Make predictions using testing data
y_pred = model.predict(X_test)

# 7. Evaluate model performance
accuracy = accuracy_score(y_test, y_pred)

print("\nMODEL ACCURACY:")
print(round(accuracy * 100, 2), "%")

print("\nCLASSIFICATION REPORT:")
print(classification_report(y_test, y_pred))

# 8. Create a confusion matrix
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=['Negative', 'Neutral', 'Positive']
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='RdPu',
    xticklabels=['Negative', 'Neutral', 'Positive'],
    yticklabels=['Negative', 'Neutral', 'Positive']
)

plt.title('Logistic Regression Confusion Matrix')
plt.xlabel('Predicted Sentiment')
plt.ylabel('Actual Sentiment')
plt.tight_layout()
plt.show()
```

```python

# STEP 14: INTERPRETATION OF LOGISTIC REGRESSION RESULTS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)

# 1. Calculate overall accuracy
accuracy = accuracy_score(y_test, y_pred)

print("=" * 55)
print("LOGISTIC REGRESSION MODEL INTERPRETATION")
print("=" * 55)

print(f"\nOverall Model Accuracy: {accuracy * 100:.2f}%")

# 2. Calculate precision, recall and F1-score
labels = ['Negative', 'Neutral', 'Positive']

precision, recall, f1, support = precision_recall_fscore_support(
    y_test,
    y_pred,
    labels=labels,
    zero_division=0
)

results = pd.DataFrame({
    'Sentiment': labels,
    'Precision': precision,
    'Recall': recall,
    'F1-Score': f1,
    'Number of Reviews': support
})

print("\nPERFORMANCE BY SENTIMENT CATEGORY:")
print(results.round(3).to_string(index=False))

# 3. Identify strongest and weakest categories
best_category = results.loc[
    results['F1-Score'].idxmax(), 'Sentiment'
]

weakest_category = results.loc[
    results['F1-Score'].idxmin(), 'Sentiment'
]

print("\nMODEL PERFORMANCE INTERPRETATION:")
print(f"The model performed best when identifying {best_category} reviews.")
print(f"The model struggled most when identifying {weakest_category} reviews.")

# 4. Plot performance comparison
results_plot = results.set_index('Sentiment')[
    ['Precision', 'Recall', 'F1-Score']
]

results_plot.plot(
    kind='bar',
    figsize=(11, 6),
    color=['#c71585', '#e75480', '#ff69b4']
)

plt.title('Logistic Regression Performance by Sentiment')
plt.xlabel('Sentiment Category')
plt.ylabel('Performance Score')
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(title='Evaluation Metrics')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 5. Display confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=labels)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='RdPu',
    xticklabels=labels,
    yticklabels=labels
)

plt.title('Confusion Matrix: Actual vs Predicted Sentiment')
plt.xlabel('Predicted Sentiment')
plt.ylabel('Actual Sentiment')
plt.tight_layout()
plt.show()

# 6. Calculate correctly and incorrectly classified reviews
correct = (y_test == y_pred).sum()
incorrect = (y_test != y_pred).sum()

print("\nPREDICTION SUMMARY:")
print("Correct predictions:", correct)
print("Incorrect predictions:", incorrect)

print("\nFINAL CONCLUSION:")
print(f"The model achieved {accuracy * 100:.2f}% accuracy.")
print(f"Its strongest category was {best_category}.")
print(f"Its weakest category was {weakest_category}.")
print("Further improvements may help the model classify weaker categories.")
```

```python

# STEP 15: IMPROVING AND RETRAINING THE MODEL

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    precision_recall_fscore_support,
    confusion_matrix
)

# ------------------------------------------------
# 1. Create an improved Logistic Regression model
# ------------------------------------------------

improved_model = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=30000,       # Increase vocabulary size
        ngram_range=(1, 2),       # Use single words and word pairs
        min_df=2,                 # Ignore very rare words
        max_df=0.95,               # Ignore extremely common words
        sublinear_tf=True,         # Reduce the effect of repeated words
        stop_words='english'
    )),

    ('classifier', LogisticRegression(
        C=2.0,                     # Adjust regularisation
        max_iter=2000,             # Allow more training iterations
        class_weight='balanced',   # Give more attention to minority classes
        solver='lbfgs',
        random_state=42
    ))
])

# ------------------------------------------------
# 2. Train the improved model
# ------------------------------------------------

print("Training the improved Logistic Regression model...")

improved_model.fit(X_train, y_train)

print("Improved model training completed!")

# ------------------------------------------------
# 3. Make predictions
# ------------------------------------------------

improved_predictions = improved_model.predict(X_test)

# ------------------------------------------------
# 4. Calculate improved model accuracy
# ------------------------------------------------

improved_accuracy = accuracy_score(
    y_test,
    improved_predictions
)

print("\nIMPROVED MODEL ACCURACY:")
print(round(improved_accuracy * 100, 2), "%")

# ------------------------------------------------
# 5. Display classification report
# ------------------------------------------------

print("\nIMPROVED MODEL CLASSIFICATION REPORT:")

print(classification_report(
    y_test,
    improved_predictions,
    labels=['Negative', 'Neutral', 'Positive'],
    zero_division=0
))

# ------------------------------------------------
# 6. Create confusion matrix
# ------------------------------------------------

labels = ['Negative', 'Neutral', 'Positive']

cm = confusion_matrix(
    y_test,
    improved_predictions,
    labels=labels
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='RdPu',
    xticklabels=labels,
    yticklabels=labels
)

plt.title('Improved Logistic Regression Confusion Matrix')
plt.xlabel('Predicted Sentiment')
plt.ylabel('Actual Sentiment')
plt.tight_layout()
plt.show()

# ------------------------------------------------
# 7. Compare original and improved models
# ------------------------------------------------

original_precision, original_recall, original_f1, _ = (
    precision_recall_fscore_support(
        y_test,
        y_pred,
        labels=labels,
        zero_division=0
    )
)

new_precision, new_recall, new_f1, _ = (
    precision_recall_fscore_support(
        y_test,
        improved_predictions,
        labels=labels,
        zero_division=0
    )
)

comparison = pd.DataFrame({
    'Sentiment': labels,
    'Original F1-score': original_f1,
    'Improved F1-score': new_f1
})

print("\nMODEL COMPARISON:")
print(comparison.round(3).to_string(index=False))

# ------------------------------------------------
# 8. Compare overall accuracy
# ------------------------------------------------

original_accuracy = accuracy_score(y_test, y_pred)

accuracy_comparison = pd.DataFrame({
    'Model': ['Original Model', 'Improved Model'],
    'Accuracy (%)': [
        original_accuracy * 100,
        improved_accuracy * 100
    ]
})

print("\nACCURACY COMPARISON:")
print(accuracy_comparison.round(2).to_string(index=False))

# ------------------------------------------------
# 9. Visualise model comparison
# ------------------------------------------------

comparison.set_index('Sentiment').plot(
    kind='bar',
    figsize=(10, 6),
    color=['#c71585', '#ff69b4']
)

plt.title('Original vs Improved Model: F1-score Comparison')
plt.xlabel('Sentiment Category')
plt.ylabel('F1-score')
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print("\nMODEL IMPROVEMENT ANALYSIS COMPLETED!")
```
